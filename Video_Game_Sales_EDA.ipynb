1. Load Data Set

```python
import pandas as pd
import numpy as np

# 1. Load the dataset
file_path = '/Users/shehram/Downloads/Video_Games_Sales_as_at_22_Dec_2016.csv'
df = pd.read_csv(file_path)
```

2. Display First Ten Rows

```python
print("--- First Ten Rows ---")
print(df.head(10))
```

3. Find Dimensions

```python
# The first number is rows (total games), the second is columns (attributes)
print(f"\nDimensions (Rows, Columns): {df.shape}")
```

4. Top Five Games by Critic Score

```python
top_5_critic = df.sort_values(by='Critic_Score', ascending=False).head(5)
print("\n--- Top 5 Games by Critic Score ---")
print(top_5_critic[['Name', 'Critic_Score']])
```

5. Number of Games in Each Genre

```python
print("\n--- Games per Genre ---")
print(df['Genre'].value_counts())
```

6. First Five Games on SNES Platform

```python
snes_games = df[df['Platform'] == 'SNES'].head(5)
print("\n--- First Five SNES Games ---")
print(snes_games[['Name', 'Platform']])
```

7. Top Five Publishers by Total Global Sales

```python
top_publishers = df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False).head(5)
print("\n--- Top 5 Publishers by Global Sales ---")
print(top_publishers)
```

8. New column: Percentage of Global Sales From North America

```python
df['NA_Sales_Percentage'] = (df['NA_Sales'] / df['Global_Sales']) * 100
print("\n--- First Five Rows with NA Sales Percentage ---")
print(df[['Name', 'NA_Sales', 'Global_Sales', 'NA_Sales_Percentage']].head(5))
```

9. Number of NaN (missing) Entries Per Column

```python
print("\n--- Missing Values per Column ---")
print(df.isna().sum())
```

10. Handle User_Score (convert 'tbd' to NaN, calculate median, and fill)

```python
# errors='coerce' turns non-numeric strings like 'tbd' into NaN automatically
df['User_Score'] = pd.to_numeric(df['User_Score'], errors='coerce')

# Calculate the median
median_user_score = df['User_Score'].median()
print(f"\nMedian User Score: {median_user_score}")

# Fill all NaN entries in User_Score with that median
df['User_Score'] = df['User_Score'].fillna(median_user_score)
print("\n--- First Five User Scores (After Filling NaNs) ---")
print(df['User_Score'].head(5))
```
